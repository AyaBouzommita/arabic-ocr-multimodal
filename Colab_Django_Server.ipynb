### 🔑 Setup Ngrok & Run Server
1. Go to [dashboard.ngrok.com](https://dashboard.ngrok.com/) and copy your Authtoken.
2. Paste it in the prompt when running the cell below.

```python
import getpass
from pyngrok import ngrok, conf
import os
import threading

# Enter Ngrok Token
print("Enter your ngrok authtoken (from dashboard.ngrok.com):")
NGROK_TOKEN = getpass.getpass()
conf.get_default().auth_token = NGROK_TOKEN

# Set Gemini API Key for the formatting reconstructor
print("\nEnter your Gemini API Key (for Markdown formatting):")
os.environ['GEMINI_API_KEY'] = getpass.getpass()

# Create tunnel
public_url = ngrok.connect(8000).public_url
print(f"\n🌍 YOUR WEBSITE IS LIVE AT: {public_url} 🌍\n")

# Run Django Server
!python neoledge_web/manage.py runserver 0.0.0.0:8000
```
